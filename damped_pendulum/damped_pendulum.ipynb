# Damped Pendulum Simulator

An interactive damped pendulum built with `numpy`, `matplotlib` and `ipywidgets` sliders.
Drag the sliders and the angle-vs-time plot redraws immediately.

This is the Python counterpart of a MATLAB Live Script with slider controls
(or a figure with `uicontrol('Style', 'slider')` callbacks).

Requires a Jupyter front end with widget support: `pip install -r ../requirements.txt`, then `jupyter lab`.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
```

## Model

The pendulum is modelled with the small-angle, closed-form solution of a damped oscillator:

$$\theta(t) = \theta_0 \, e^{-c\,t} \cos(\omega t), \qquad \omega = \sqrt{\frac{g}{L}}, \qquad g = 9.81\ \mathrm{m/s^2}$$

| Slider    | Symbol     | Meaning                        |
|-----------|------------|--------------------------------|
| `length`  | $L$        | pendulum length in metres      |
| `damping` | $c$        | exponential decay rate in 1/s  |
| `angle0`  | $\theta_0$ | initial angle in degrees       |

This is an idealised model: the envelope decays at exactly the slider's damping rate and the
oscillation uses the undamped natural frequency $\sqrt{g/L}$. It is not a numerical solution of
the full nonlinear pendulum equation, so treat it as a teaching demo rather than a precision tool.

```python
def pendulum(length=1.0, damping=0.1, angle0=30):
    t = np.linspace(0, 10, 500)
    omega = np.sqrt(9.81 / length)
    theta = np.radians(angle0) * np.exp(-damping * t) * np.cos(omega * t)
    plt.figure(figsize=(6, 3))
    plt.plot(t, np.degrees(theta))
    plt.xlabel("Time (s)")
    plt.ylabel("Angle (deg)")
    plt.title(f"Pendulum: L={length}m, damping={damping}")
    plt.grid(True)
    plt.show()
```

## Interactive plot

Move the sliders to change the pendulum. Each change calls `pendulum(...)` again and redraws the figure.

```python
interact(
    pendulum,
    length=FloatSlider(min=0.1, max=3.0, step=0.1, value=1.0),
    damping=FloatSlider(min=0.0, max=1.0, step=0.05, value=0.1),
    angle0=FloatSlider(min=1, max=90, step=1, value=30),
)
```
